```python
import torch
```

```python
# To create a tensor, just put a list
a = torch.tensor([1, 2, 3])
print(a)

# Shape is 3: we have a vector of 3 values

print(a.ndim) #1

print(a.dtype) #uint 64.

# Torch looks at the value: if any of the values were a floating point, the type will be float32

# Can specify dtype: a = torch.tensor([1, 2, 3], dtype=torch.float32)

# Creating a tensor for a large number of values: tedius! Need to provide a large list of values, and convert it to a tensor
```

```python
# Idea: create a tensor with only zeros

b = torch.zeros(3, dtype=torch.float32)
print(b)

# We can add as many zeroes as possible

b2 = torch.zeros(10, dtype=torch.float32)
print(b2)

b3 = torch.zeros([10, 10], dtype=torch.float32)
print(b3)

b4 = torch.zeros([3, 3, 3], dtype=torch.float32)
print(b4)

b5 = torch.ones([10, 10], dtype=torch.float32)
print(b5)

b6 = torch.rand([3, 3, 3], dtype=torch.float32)
print(b6)

# 3 types normally: zeros, ones, and random values
```

```python
# We can load tensors from images

from PIL import Image

img = Image.open('cat.jpg')

# PIL image: have to convert it to a tensor

# torch.as_tensor(img) # will not work directly: cannot infer dtype of JpegImageFile

# Pytorch: converts numpy arrays into tensors but does not convert images

import numpy as np
c = torch.as_tensor(np.array(img))

# Now, instead of the image, we see the RGB values, stored in a 3-dimensional tensor.

print(c.shape)
# height, width, number of channels

# dtype: uint8
```

```python
# Basic operations for tensors

a = torch.arange(10)
b = torch.ones(10)
print(a)
print(b)

print(a + b) # add two tensors
# Can perform multiplication, subtraction, division, power

print(a ** (2*b))
```

```python
# Manipulating tensors and their shapes

a = torch.arange(6)
print(a)
print(a.shape)
print(a.view(2, 3)) # Keep the values around, but change the shape
# 2 functions here: view and reshape

# View: does not copy the underlying data: simply changes the shape
# Reshape: copies the underlying data
```

```python
# Next: basic linear algebra

a = torch.rand(2, 3)
print(a)
print(a.mT) # matrix transpose.
```

```python
a = torch.rand(2, 3, 4)
print(a) # 2 matrices of lists 3 * 4

# Permute: swaps any dimension
print(a.permute(1, 2, 0).shape)
print(a.permute(1, 2, 0))
# data in the 2nd dimension is now stored in the first dimension
# Data in the 3rd dimension is now stored in the second dimension
# data stored in the 1st dimension is now in the last dimension

# Helpful for loading images or other data that is rearranged, and helps such that deep networks can process it
# In the input to the deep network, you will see permutations of your data that you can feed in to the network
```

```python
# Add, subtract dimensions

a = torch.arange(6)
print(a[None])
print(a[None].shape)


# Creates a singleton dimension, so we add one of size one, where we specify a none


# Add the None to the end:
print(a[:, None].shape)

# Add a None to the start and end of the data
print(a[None, :, None].shape)

# Colon: your first sort of dimension that you're looking at

# Additional dimensions of size one can be viewed anywhere

print(a[..., None].shape) # ... if you don't know the number of dimensions

print(a[..., None, :, None].shape) # : is the last dimension
```

```python
# We can remove dimensions from a tensors as long as it is of size one

a = torch.arange(6).view(3, 2, 1, 1)
print(a[..., 0].shape)

print(a.squeeze(-1).shape) # Remove last dimension
print(a.squeeze(-2).shape) # remove second last dimension

# Cannot squeeze first dimension, because it is not size one

# Squeeze with no arguments: remove all dimensions
# Dangerous! Never use it. You may remove data that you would want.
```

```python
# Specific concept on how to add tensors that don't quite have the same shape.

# a = torch.rand(4)
# b = torch.rand(5)

# print(a + b) # error

# If mismatch is between dimensions of size 1, we can add tensors together

a = torch.rand(4)
b = torch.rand(1)

print(a + b) # this works

# Interesting concept: we can increase the dimensions of the tensor

a = torch.rand((4, 1))
b = torch.rand((1, 5))
print(a + b) # this works

# This is known as broadcasting: what PyTorch does internally, it says to take a tensor of size 4 * 1, and add a 1 * 5 tensor to it
# Replicate this above tensor, replicate it 5 times
# replicate bottom one 4 times, and add them together

# Powerful concept: complex mathematical calculations are in a small form

a = torch.arange(4).view(4, 1)
b = torch.arange(5).view(1, 5) * 10

print(f'{a = }')
print(f'{b = }')
print(a * b) # This is like the outer product

# Create a big output matrix like this

# Broadcasting allows us to express any operation that we can fit into the right shapes
# Any element-wise operation applies to all possible elements
```

```python
x = torch.randn(10, 2)
d = torch.zeros(10, 10)

# distance between all pairs of random numbers

# without broadcasting:
for i in range(10):
  for j in range(10):
    d[i, j] = (x[i] - x[j]).pow(2).sum()
print(d) # Diagonal is zero
# This is not very efficient, tedious to write and not the easiest to read


# Using broadcasting:
# add a singleton dimension, take x, add the one in the first position
# This subtracts any row of X from any other row of x
# Take the suqre elements, and sum along the last dimension (the dimension along the channels of X)

d = (x[:, None, :] - x[None, :, :]).pow(2).sum(-1) # Compact way to write complex notation
print(d)
```

```python
x = torch.randn(10, 2)
d = torch.zeros(10, 10)

# Now: want the maximum distance
max_dist, max_idx = 1e100, (-1, -1)
# without broadcasting:
for i in range(10):
  for j in range(10):
    if (x[i] - x[j]).pow(2).sum() > max_dist:
      max_dist, max_idx = (x[i] - x[j]).pow(2).sum(), (i, j)
print(max_dist, max_idx)

# In pytorch: compute the pairwise distance using broadcasting, then compute the maximum value and the argmax out of it

d = (x[:, None, :] - x[None, :, :]).pow(2).sum(-1) # Compact way to write complex notation
print(d.max(), (d.argmax() // 10, d.argmax() % 10))

# Caveat: argmax flattens the array first, and then computes the distance


# using tensors is a lot easier

```

```python
# Examples of matrices

a = torch.rand(2, 4)
b = torch.rand(4, 3)
c = a @ b # Matrix multiplication, like numpy
print(c.shape) # 2 * 3

# Same concept with vectors

a = torch.rand(2, 4)
b = torch.rand(4)
c = a @ b
print(c.shape) # 2
d = b @ a.mT
print(d.shape)

# norm
print(b.pow(2).sum().sqrt())
print(torch.linalg.norm(b))


#Pytorch: allows you to create tensors, manipulate them, reshape them, combine them in concise notation
```
